#### Visualising digital signal test

```python
size = 20

x = [0]
y = []

for i in range(1, size):
    if i % 2 != 0:
        x.append(x[i-1] + 1)
    else:
        x.append(x[i-1])

for j in range(size // 2):
    if j % 2 == 0:
        y.append(1)
        y.append(1)
    else:
        y.append(0)
        y.append(0)


print(len(y))
print(len(x))

plt.figure(figsize=[7,7])
plt.plot(x, y)

plt.show()

    
```

### Approximation with sines

```python
x = [0, np.pi , np.pi , 2 * np.pi]
y = [1, 1, -1, -1]
plt.plot(x, y)

inputs = np.linspace(0, 2 * np.pi, 1000)
outputs = np.sin(inputs)
plt.plot(inputs, outputs)


plt.show()
```

```python
def func(x):
    result = 0
    for i in range(0,100):
        if i % 2 != 0:
            result += np.sin(i * x) / i
    
    return result

inputs = np.linspace(0, 8 * np.pi, 1000)
outputs = func(inputs)

plt.figure(figsize=[9,4])
plt.plot(inputs, outputs)

plt.grid()


plt.show()
```
