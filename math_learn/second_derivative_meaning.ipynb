```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
inputs = np.linspace(-3,3, 5000, endpoint=False)

function = inputs**3 - 8 * inputs

first_derivative = 3 * inputs**2 - 8

second_derivative = 6 * inputs
```

```python
plt.plot(inputs, function, label='Function: $x^3 - 8x$')
plt.plot(inputs, first_derivative, label='First Derivative: $3x^2 - 8$')
plt.plot(inputs, second_derivative, label='Second Derivative: $6x$')

plt.title('Function and its Derivatives')
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black', lw=0.5, ls='--')
plt.axvline(0, color='black', lw=0.5, ls='--')
plt.legend()    

plt.grid()

plt.show()
```
