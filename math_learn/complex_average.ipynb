```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.linspace(0, 2 * np.pi, 1000)
outputs = np.exp(-1j * 2 * t)

fig, axs = plt.subplots(figsize=(5, 8), subplot_kw={'projection': 'polar'},
                        layout='constrained')

axs.plot(np.angle(outputs), np.abs(outputs), label="e^(it)")
```

```python
np.mean(outputs)
```

#### Why mean of complete rotation is 0

Mean of complete rotation is zero becasue horzontal range is [- radius, + radius], so for every horizontal point you have 2 vertical values with same magnitudes and opposite directions so whole sum, and some at any horizonatl point is 0. This also can be explain when we looks at xy formula. There are two equations for y:
- y(x) = sqrt(r^2 - x^2)
- y(x) = - sqrt(r^2 - x^2)
That's why they cancel

```python
x = np.linspace(-3, 3, 3000)

y1 = np.sqrt(9 - x**2)
y2 = - np.sqrt(9 - x**2)

plt.figure(figsize=[6,6])
plt.plot(x, y1)
plt.plot(x, y2)


plt.grid()
plt.show()
```
