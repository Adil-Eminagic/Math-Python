## How do Fourier transforms work

So in Fourier transform we are given singal in certain period - signal in time domain ( x axis time, y axis air pressure), and we need to find individual frequencies of indiviudal sine functions it consists of. It is done by converting function to from time domain to frequency domain. To make it easier in the beginning, it is easier to already define function in time domain.

- Fourier trasnform formula is **F(omega) = integral (- infinity, + infinity) f(t) * e^-j * omeaga * t**
- For finite time interval is **F(omega) 1 / t2-t1 * integal (t1, t2) f(t) * e^-j * omeaga * t**

When we definite integral by it's range, we get average value of that range. And since in complex plane there are 2 dimensions, average will also have 2 dimensions. It is mean of all points in complex plane of product of time singal function and Eulers equation. Smaller time step is, more points we include, therefore more accurate positio of mean we get

This formula also tells us that we calcuate average position, center of mass of all point that polar equation f(t) * e^-j * omeaga * t produce where t is in interval (t1,t2)

Before making graph, it is better to first do calculations

```python
# first define input values for polar equation

inputs = np.linspace(0, 2 * np.pi, 1000)
```

### Defining fourier integral

- **omega** is input, so I need function that takes omega as input
- I have alreay defined all inputs in given range, so instad of range I need list of **inputs** which I'll name them *time_steps*
- Then I need to calculate average of all those points from polar equation, but first I need their sum, and then devide it by number of points

```python
def fourier_integral(omega, time_steps):
    sum = 0

    for step in time_steps:
        sum += np.sin(3 * step) * np.exp(-1j * omega * step) # step is current time
    
    return sum / 1000

    # I don't need to store in variables becasue it values doen't repeat a lot
```

### Calculating values of function

Now when I have defined Fourier integral, I have equation for F(omega), where omega is input value. I need to calculate it for every omeaga from 0 to some value. It should go to infinity, but for this actual calculation, I need some finite interval. But maybe I don't need continous range of values, I think that I only look for integer values of omega which corresponds to frequencies of sine functions in time domain function. I need to check it. But since it is function, let's choose continoius values. So I need:

- Range of omega inputs
- To calculate output for each input in F(omega)

```python
omega_inputs = np.linspace(0, 6, 6001)

outputs = [fourier_integral(o, inputs) for o in omega_inputs]
```

```python
# probably magnitude determins value actual output of F(omega function)

print(np.abs(4 + 3j))
```

```python
plt.plot(omega_inputs, np.abs(outputs))
plt.title("Fourier Transform of sin(3t)")
plt.xlabel("Frequency (omega)")
plt.ylabel("Magnitude")
plt.show()
```
