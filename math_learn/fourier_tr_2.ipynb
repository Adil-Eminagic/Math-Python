```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define a time domain 

# More accurate calcuation require biger time interval

T = 16 * 2 * np.pi
N = 16000

time_samples = np.linspace(0, T, N, endpoint=False)
```

```python
# Define a singla

signal = 2 *np.sin(4 * time_samples) + 3 *np.sin(2 * time_samples)
```

```python
dt = time_samples[1] - time_samples[0]

def fourier_transform(omega):
    values = signal * np.exp(-1j * omega * time_samples)

    return np.sum(values) * dt
```

```python
# 4. Test different angular frequencies

omega_inputs = np.linspace(0, 6, 6000)

outputs = [fourier_transform(o) for o in omega_inputs]
```

```python
plt.plot(omega_inputs, np.abs(outputs))
plt.title("Fourier Transform of sin(3t)")
plt.xlabel("Frequency (omega)")
plt.ylabel("Magnitude")
plt.show()
```
