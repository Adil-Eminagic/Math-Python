## Computing integrals without finding antiderivative

Integrals can be computed wihtout finding antiderivatives. Integral is just sum where dx approaches to infinity. Let's pick inegral of 2x dx. We know that it is antiderivative is x^2. From geometry I know that area under curve is 1/2 * 4 * 8 = 16. From antiderivative I also know that x^2 is antiderivative, so area is 4^2 = 16. But assume i know none of these, only funtion 2x, I would pick step dx, and calcaulte avearege between every dx, and add them. So if dx = 1, I would caclualte:
- (2 * 1 - 2 * 0)/ 1 = 2
- (2 * 2 - 2 * 1)/ 1 = 2
- (2 * 3 - 2 * 2)/ 1 = 2
- (2 * 4 - 2 * 2)/ 1 = 2

And then add all those values resulting in 2 * 4 = 8... NO!!! This is slope, not area under interval. Instead f(0) * dx + f(dx) * dx + f(dx^2) * dx

- 2 * 0 * 1 = 0 
- 2 * 1 * 1 = 2
- 2 * 2 * 1 = 4
- 2 * 3 * 1 = 8 

So for dx = 1, which is huge step, area undr curve is A = 12. Lets wirete formula for computing sums and letting dx be variable. For small choice of dx = 0.01 A = 15.96, whcih is very close to 16. So that's computatioal way to calculate area under curve, which by hand is not practical, but with compter it is. So with computer you don't need to find antiderivatve in order to calcualte area under curve.

Now let's pick function x^2, to test it. It's antiderivaitve is 1/3 x^3. So area under curve from 0-4 is 4^3/3 = 64 / 3 = 21.333333. For dx = 10^-4, it is 21.33253, so it works.

```python
inputs =  np.linspace(0, 4, 10)
outputs = [2 * i for i in inputs]

plt.plot(inputs, outputs)
plt.title('Function 2x', fontsize=14)

plt.show()
```

```python
dx = 0.01

x1 = 0
x2 = 4

span = x2 - x1

steps = int(span / dx)

inputs = np.linspace(x1, x2, steps + 1)

inputs = inputs[:-1]

# print(steps)
# print(inputs)

sum = 0

for i in inputs:
    sum += 2 * i * dx

print(f"Area is {sum}")
```
