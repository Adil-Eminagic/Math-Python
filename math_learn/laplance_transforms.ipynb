```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
t = np.linspace(0, 8 * 2 * np.pi, 8000)
s =  -1 - 1j
outputs = np.cos(t) * np.exp(-s * t)

fig, axs = plt.subplots(figsize=(6,6), subplot_kw={'projection': 'polar'},
                        layout='constrained')

axs.set_title("Polar Plot of Laplace Transform Output", va='bottom')

axs.plot(np.angle(outputs), np.abs(outputs), label="e^(it)")
```

```python
for i in range(2000, 4000, 100):
    print(f'At time {round(t[i], 2)} angle is {round(np.angle(outputs[i]), 2)} and radius is {round(np.abs(outputs[i]), 2)}')
```

```python
dx = t[1] - t[0]
sum = np.sum(outputs * dx)
print(f'The sum of the outputs is: {sum}')
```
