```python
import sympy as sp
from sympy import symbols
```

```python
x = sp.symbols('x')
```

```python
f = x**3 - 6*x**2 + 11*x - 6
display(f)
```

```python
display(sp.Eq(sp.Function('f')(x), f))
```

```python
sp.diff(f, x)
```

```python
x, y = symbols('x y')
expr = x + 2*y
expr
```

```python
sp.solve(f, x)
```

```python
f.subs({x: 2})
```

```python
t, s = sp.symbols('t s', positive=True)

f = sp.exp(-2*t)

result = sp.laplace_transform(f, t, s, noconds=True)

display(result)
```

```python
t, s = sp.symbols('t s', positive=True)
f = sp.Function('f')

laplace = sp.Integral(
    f(t) * sp.exp(-s*t),
    (t, 0, sp.oo)
)

display(
    sp.Eq(
        sp.LaplaceTransform(f(t), t, s),
        laplace
    )
)
```
